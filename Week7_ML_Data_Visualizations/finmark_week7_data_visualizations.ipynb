```python
trans_final["Year"] = pd.to_datetime(trans_final["TransactionDate"]).dt.year

yearly_transactions = trans_final["Year"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

sns.lineplot(
    x=yearly_transactions.index,
    y=yearly_transactions.values,
    marker="o"
)

plt.title("Number of Transactions by Year")
plt.xlabel("Year")
plt.ylabel("Number of Transactions")

plt.show()
```

## Key Findings and Interpretations

### Numerical Findings
- Customer ages range from 18 to 70 years, with a mean of approximately 44.69 years and a median of 45 years.
- Transaction amounts range from 0.00 to 999.86, with a mean of approximately 493.62 and a median of 498.015.
- Transaction amounts show noticeable variability, with a standard deviation of approximately 279.25.

### Customer Demographics
- The customer age histogram shows customers distributed across the 18–70 age range, with a pronounced concentration around age 45.
- Female and male customers have relatively similar representation, with female customers slightly more numerous.
- High, Low, and Medium income levels have relatively similar counts, while the Unknown category has considerably fewer records.

### Transaction Patterns
- Payment methods have relatively similar transaction counts, with Debit Card having the highest count and Bank Transfer having the lowest.
- The named product categories have relatively similar transaction counts, while the Unknown category has substantially fewer records.
- The transaction amount histogram shows a pronounced concentration around 500. This concentration is partly influenced by median imputation used during data cleaning.
- The transaction amount box plot does not show obvious outliers beyond the whiskers.

### Social Media Patterns
- Instagram, Twitter, and Facebook have relatively similar numbers of recorded interactions.
- Positive, Neutral, and Negative sentiments account for most recorded interactions and have relatively similar frequencies.
- Very Negative and Very Positive sentiments occur much less frequently than the other sentiment categories.

### Time Pattern
- The number of recorded transactions was highest in 2023, while 2022 and 2024 had substantially fewer recorded transactions.
- Because the finalized transaction data has uneven monthly coverage across years, this difference should be interpreted as a difference in recorded transactions rather than definitive evidence of a business trend.

### Relevance to Future Machine Learning
- The numerical variables Age and Transaction Amount provide measurable features that may be useful for future analysis.
- Categorical variables such as Gender, Income Level, Product Category, Payment Method, Platform, and Sentiment may also provide useful features.
- Categories with relatively few observations, such as Very Negative and Very Positive sentiment, may require consideration when preparing data for future machine learning tasks.
