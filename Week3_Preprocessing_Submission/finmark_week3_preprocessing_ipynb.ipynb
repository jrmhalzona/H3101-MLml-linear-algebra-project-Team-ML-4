```python
import pandas as pd

transactions = pd.read_csv(
    '/content/customer_transactions_preprocessed.csv'
)

social_media = pd.read_csv(
    '/content/social_media_interactions_preprocessed.csv'
)

demographics = pd.read_csv(
    '/content/customer_demographics_preprocessed.csv'
)

print("Preprocessed datasets loaded successfully!")
print("Transactions:", transactions.shape)
print("Social Media:", social_media.shape)
print("Demographics:", demographics.shape)
```

```python
print("=" * 60)
print("CUSTOMER TRANSACTIONS")
print("=" * 60)

print("Rows and columns:", transactions.shape)
print("Duplicate rows:", transactions.duplicated().sum())
print("Missing values:")
print(transactions.isna().sum())
print("\nData types:")
print(transactions.dtypes)


print("\n" + "=" * 60)
print("SOCIAL MEDIA INTERACTIONS")
print("=" * 60)

print("Rows and columns:", social_media.shape)
print("Duplicate rows:", social_media.duplicated().sum())
print("Missing values:")
print(social_media.isna().sum())
print("\nData types:")
print(social_media.dtypes)


print("\n" + "=" * 60)
print("CUSTOMER DEMOGRAPHICS")
print("=" * 60)

print("Rows and columns:", demographics.shape)
print("Duplicate rows:", demographics.duplicated().sum())
print("Missing values:")
print(demographics.isna().sum())
print("\nData types:")
print(demographics.dtypes)
```

```python
# Convert date columns to proper datetime format

transactions['TransactionDate'] = pd.to_datetime(
    transactions['TransactionDate'],
    errors='coerce',
    dayfirst=True
)

social_media['InteractionDate'] = pd.to_datetime(
    social_media['InteractionDate'],
    errors='coerce',
    dayfirst=True
)

demographics['SignupDate'] = pd.to_datetime(
    demographics['SignupDate'],
    errors='coerce',
    dayfirst=True
)

print("✅ Date columns converted successfully!")
```

```python
print("Transactions:", transactions['TransactionDate'].dtype)
print("Social Media:", social_media['InteractionDate'].dtype)
print("Demographics:", demographics['SignupDate'].dtype)
```

```python
print("=" * 60)
print("FINAL VERIFICATION")
print("=" * 60)

print("\nTRANSACTIONS")
print("Rows:", len(transactions))
print("Columns:", len(transactions.columns))
print("Duplicates:", transactions.duplicated().sum())
print("Missing values:", transactions.isna().sum().sum())
print(transactions.dtypes)

print("\nSOCIAL MEDIA")
print("Rows:", len(social_media))
print("Columns:", len(social_media.columns))
print("Duplicates:", social_media.duplicated().sum())
print("Missing values:", social_media.isna().sum().sum())
print(social_media.dtypes)

print("\nDEMOGRAPHICS")
print("Rows:", len(demographics))
print("Columns:", len(demographics.columns))
print("Duplicates:", demographics.duplicated().sum())
print("Missing values:", demographics.isna().sum().sum())
print(demographics.dtypes)
```

```python
# Reload the Social Media preprocessed CSV from the file
social_media = pd.read_csv(
    '/content/social_media_interactions_preprocessed.csv'
)

# Convert the existing dates correctly
social_media['InteractionDate'] = pd.to_datetime(
    social_media['InteractionDate'],
    errors='coerce'
)

print("Social Media restored and dates converted!")

print("\nRows:", len(social_media))
print("Duplicates:", social_media.duplicated().sum())
print("Missing values:", social_media.isna().sum().sum())
print("\nMissing by column:")
print(social_media.isna().sum())
print("\nDate type:")
print(social_media['InteractionDate'].dtype)
```

```python
print("=" * 60)
print("FINAL SUBMISSION CHECK")
print("=" * 60)

print("\nCUSTOMER TRANSACTIONS")
print("Rows:", len(transactions))
print("Columns:", len(transactions.columns))
print("Duplicates:", transactions.duplicated().sum())
print("Missing values:", transactions.isna().sum().sum())
print(transactions.dtypes)

print("\nSOCIAL MEDIA INTERACTIONS")
print("Rows:", len(social_media))
print("Columns:", len(social_media.columns))
print("Duplicates:", social_media.duplicated().sum())
print("Missing values:", social_media.isna().sum().sum())
print(social_media.dtypes)

print("\nCUSTOMER DEMOGRAPHICS")
print("Rows:", len(demographics))
print("Columns:", len(demographics.columns))
print("Duplicates:", demographics.duplicated().sum())
print("Missing values:", demographics.isna().sum().sum())
print(demographics.dtypes)
```

```python
# Save the FINAL preprocessed datasets

transactions.to_csv(
    '/content/customer_transactions_preprocessed.csv',
    index=False
)

social_media.to_csv(
    '/content/social_media_interactions_preprocessed.csv',
    index=False
)

demographics.to_csv(
    '/content/customer_demographics_preprocessed.csv',
    index=False
)

print("✅ FINAL PREPROCESSED DATASETS SAVED!")
```
